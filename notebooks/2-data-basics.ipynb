**Question:** how much known order value is recorded for each category? The six records below are entirely fictional. Each row should be one order. A blank amount means we do not know it, not that it is zero.


```python
import pandas as pd
import plotly.express as px

orders = pd.DataFrame([
    {"order_id": "A1", "category": "Books", "value_eur": 10},
    {"order_id": "A2", "category": "Books", "value_eur": 20},
    {"order_id": "A3", "category": "Games", "value_eur": 30},
    {"order_id": "A4", "category": "Games", "value_eur": 40},
    {"order_id": "A5", "category": "Books", "value_eur": None},
    {"order_id": "A1", "category": "Books", "value_eur": 10},
])
orders
```

## 1. Inspect before changing anything

A **duplicate** is a repeated row. A **missing value** is information we do not have. Find both in the displayed table before running the next cell.


```python
print("Rows:", len(orders))
print("Exact duplicate rows:", orders.duplicated().sum())
print("Missing amounts:", orders["value_eur"].isna().sum())
```

## 2. Make one justified transformation

We know A1 was accidentally copied, so remove its exact duplicate. Keep the missing amount blank: inventing a sale would change the meaning. In real data, repeated rows are not automatically mistakes; ask what one row represents.


```python
clean = orders.drop_duplicates().copy()
print("Rows after removing the duplicate:", len(clean))
print("Known value before:", orders["value_eur"].sum())
print("Known value after:", clean["value_eur"].sum())
clean
```

## 3. Summarize, then visualize

**Group by** collects rows with the same category. A **sum** adds their known amounts. We explicitly keep only known amounts when calculating the chart. The chart is a picture of that summary, not an extra source of evidence.


```python
known = clean.dropna(subset=["value_eur"])
totals = known.groupby("category", as_index=False)["value_eur"].sum()
print(totals.to_string(index=False))
figure = px.bar(totals, x="category", y="value_eur",
                title="Known order value by category — one Books amount is missing",
                labels={"category": "Category", "value_eur": "Known order value (EUR)"})
figure.show()
```
